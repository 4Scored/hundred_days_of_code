```python
import pandas as pd
pd.options.display.float_format = '{:,.2f}'.format 
```

```python
df = pd.read_csv('salaries_by_college_major.csv')
df.head() 
```

```python
print(df.shape)
print("\n\n")
print(df.columns)
print("\n\n")
print(df.isna())
```

```python
clean_df = df.dropna()
clean_df
```

```python
print(clean_df['Starting Median Salary'])
print("\n\n")
print(clean_df['Starting Median Salary'].max())
print("\n\n")
print(clean_df['Starting Median Salary'].idxmax())
print("\n\n")
print(clean_df['Undergraduate Major'].loc[43])
print("\n\n")
print(clean_df['Undergraduate Major'][43])
print("\n\n")
print(clean_df.loc[43])
```

```python
highest_mid_career_salary = clean_df['Mid-Career Median Salary'].max()
highest_mid_career_salary_idx = clean_df['Mid-Career Median Salary'].idxmax()
print(f"Highest Mid-Career Salary is ${highest_mid_career_salary} for {clean_df['Undergraduate Major'][highest_mid_career_salary_idx]} majors")
```

```python
lowest_start_career_salary = clean_df['Starting Median Salary'].min()
lowest_start_career_salary_idx = clean_df['Starting Median Salary'].idxmin()
print(f"Lowest Starting Salary is ${lowest_start_career_salary} for {clean_df['Undergraduate Major'][lowest_mid_career_salary_idx]} majors")
```

```python
lowest_mid_career_salary = clean_df['Mid-Career Median Salary'].min()
lowest_mid_career_salary_idx = clean_df['Mid-Career Median Salary'].idxmin()
print(f"Lowest Mid-Career Salary is ${lowest_mid_career_salary} for {clean_df['Undergraduate Major'][lowest_mid_career_salary_idx]} majors")
```

```python
# clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
# clean_df['Mid-Career 90th Percentile Salary'].subtract(clean_df['Mid-Career 10th Percentile Salary'])
spread_col = clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
clean_df.insert(1, 'Spread', spread_col)
clean_df.head()
```

```python
low_risk = clean_df.sort_values('Spread')
low_risk[['Undergraduate Major', 'Spread']].head()
```

```python
highest_potential = clean_df.sort_values('Mid-Career 90th Percentile Salary', ascending=False)
highest_potential[['Undergraduate Major', 'Mid-Career 90th Percentile Salary']].head()
```

```python
greatest_spread = clean_df.sort_values('Spread', ascending=False)
greatest_spread[['Undergraduate Major', 'Spread']].head()
```
